# Filtrando Información

```python
import pandas as pd
pd.set_option('display.max_columns', None) # Ver todas las columnas [2]
pd.set_option('display.width', 1000)      # Ajustar ancho de visualización
path = "../Pandas/datasets/Online_Retail.csv"

#Recapitulando
# leer el dataset de ventas desde un archivo CSV
sales_data = pd.read_csv(path, encoding='latin-1')
#sales_data.head(3)
#sales_data.info()

# Convertir la columna 'InvoiceData' a tipo datetime
sales_data['InvoiceDate'] = pd.to_datetime(sales_data['InvoiceDate'])

# Detectar valores faltantes
missing_data = sales_data.isna().sum()
print(missing_data)

# Eliminar filas con valores faltantes en las columnas críticas
sales_data.dropna(subset='Description', inplace=True)
#sales_data.info()

# Creamos una nueva columna TotalPrice
sales_data['TotalPrice'] = sales_data['Quantity'] * sales_data['UnitPrice']
sales_data.head(5)
```

## Filtramos las ventas en el Reino Unido

```python
# Para filtrar no necesitamos utilizar ninguna función como Filter o lambda y recorre una función. NO.
# Simplemente utilizamos la misma forma de corchetes sales_data[sales_data[]== 'A algo']
uk_sales = sales_data[sales_data['Country'] == 'United Kingdom']
print(uk_sales)
```

## Filtrar por cantidades

```python
high_quantity = sales_data[sales_data['Quantity'] > 30]
#print(high_quantity.info())
print(high_quantity)
```

## Filtrado Múltiple

```python
uk_high_sales = sales_data[(sales_data['Country'] == 'United Kingdom') & (sales_data['Quantity'] > 100)]
print(uk_high_sales)
```

## Filtrando por año

```python
sales_2011 = sales_data[sales_data['InvoiceDate'].dt.year == 2011]
print(sales_2011)
```

##Filtrando por mes

```python
sales_dec_2010 = sales_data[(sales_data['InvoiceDate'].dt.year == 2011) & (sales_data['InvoiceDate'].dt.month == 1)]
print(sales_dec_2010)
```
