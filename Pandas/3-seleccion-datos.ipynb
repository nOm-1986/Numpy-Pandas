## Selección de datos
Formas de selección
iloc ==> ayuda a extraer la información del dataframe indicando el indice.
loc ===> accede a la información especificando la etiqueta.

```python
import pandas as pd
```

```python
path = "./datasets/Online_Retail.csv"
retail_data = pd.read_csv(path, encoding='latin-1')
```

```python
# Seleccionar primera fila
first_row = retail_data.iloc[0]
print(first_row)
```

```python
# Acceder a las primeras 5 filas
first_five_rows = retail_data.iloc[:5]
print(first_five_rows)
```

```python
# Utilizando un subset
subset_one = retail_data.iloc[:3,:4]
print(subset_one)

```

```python
subset_two = retail_data.iloc[2,2]
print(subset_two)
```

Seleccion de filas con loc

```python
row_index_3_loc = retail_data.loc[:,['Description', 'Quantity', 'StockCode']]
print(row_index_3_loc)
```
