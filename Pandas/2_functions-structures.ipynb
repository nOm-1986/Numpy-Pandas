```python
import pandas as pd
```

```python
path = "./datasets/Online_Retail.csv"
retail_data = pd.read_csv(path, encoding='latin-1')
columns_name = retail_data.columns
print(columns_name)
num_rows, num_colums = retail_data.shape
print(f"Filas: {num_rows} ===== Columnas: {num_colums}")
retail_data.head(4)


```

Tambien se puede obtener una descripción estadística de cada una de las columnas que tiene información cuantitativa.

```python
daily_sales = retail_data['Quantity']
print(daily_sales)
```

```python
# Resumen del data frame
summary = retail_data.describe()
print(summary)
```

```python
# Media
mean_value = daily_sales.mean()
print("La media es: ", mean_value)
```

```python
# Mediana
median_value = daily_sales.median()
print("la mediana es: ",median_value)
```

```python
# SUMA
sum_values = daily_sales.sum()
print("Total: ", sum_values)
```

```python
# Conteo de valores
count_values = daily_sales.count()
print("Conteo de valores: ", count_values)
```
