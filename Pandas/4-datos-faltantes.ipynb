## Datos faltantes - Datos Nulos
Existen 2 formas de verificar si los datos están o no. Pandas maneja 2, si el dato está disponible o si es nulo.
Para saber si es nulo utilizo el método
### isna()

```python
import pandas as pd
```

```python
path = "./datasets/Online_Retail.csv"
retail_data = pd.read_csv(path, encoding='latin-1')
retail_data.head(3)
```

```python
missing_data =  retail_data.isna()
missing_data.head(4)
```

## Contar de cada columna cuantos datos faltan
Esto se logra utilizando tambien isna() pero agregamos sum()
isna().sum()

```python
missing_data_count = retail_data.isna().sum()
print("Conteo de datos faltantes por columna:\n",missing_data_count)
```

# Qué pasa si existen datos faltantes, cómo los manejo? Debo eliminar las filas o mejor utilizo otra estrategia?
Para el ejemplo de abajo va a eliminar las columnas con datos faltantes.
PERO OJO ESTO NO ES UNA BUENA PRACTICA

```python
# axis = 1, me indica que borra la columna
no_missing_data = retail_data.dropna(axis=1)
print("Datos sin filas con valores faltantes: \n", no_missing_data)
no_missing_data.head(2)
```

# Rellenar los datos faltantes con el valor 0
## .fillna(0)

```python
# Rellenar valores con 0
retail_data_filled_zeros = retail_data.fillna(0)
sum_zero_data = retail_data_filled_zeros.isna().sum()
print(sum_zero_data)
```

## Rellenar los datos faltantes pero utilizando una técnica estadística
Ejemplo, hayando la media de la columna si es un dato numérico
* lo que debo hacer primero es calcular la media .mean()
* Después indicando la columna['tu-columna'].fillna(el-valor)

```python
mean_unit_prices = retail_data['UnitPrice'].mean()
retail_data_filled_mean = retail_data['UnitPrice'].fillna(mean_unit_prices)
print(retail_data_filled_mean)
```
