Manipulación de columnas

```python
import pandas as pd
pd.set_option('display.max_columns', None) # Ver todas las columnas [2]
pd.set_option('display.width', 1000)      # Ajustar ancho de visualización
```

```python
path = "./datasets/Online_Retail.csv"
df = pd.read_csv(path, encoding="latin-1")
df.head(2)
```

```python
df["Total_price"] = df['Quantity'] * df["UnitPrice"]
df.head(5)
```

```python
df['HighValues'] = df['Total_price'] > 20
print(df['HighValues'].head(3))
```

```python
df.info()
```

## Cambiar el formato --> Ejemplo pasar un str a un tipo datime
PAra este caso utilizamos como ejemplo la columna InvoiceDate que es de tipo str y con la función de panda to_datime() realizamos la manipulación

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
print(df.info())
```

# Cambio utilizando lambda

```python
df['DiscountedPrice'] = df['UnitPrice'].apply(lambda x: x* 0.9)
df.head(3)
```

# Aplicar una función a cada una de nuestras filas
Nuevamente utilizamos la función apply

```python
def categorize_price(price):
  if price > 20:
    return 'High'
  elif price > 10 and price <= 20:
    return 'Medium'
  else:
    return 'Low'

df['PriceCategory'] = df['Total_price'].apply(categorize_price)
df.head(10)
```
