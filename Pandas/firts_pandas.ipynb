```python
# creación de datafreme apartir de datos sencillos
data = np.arange(1,10).reshape(3,3)
dt_from_array = pd.DataFrame(data, columns=['A','B','C'])
print(dt_from_array)
```

## Creación de Dataframe desde una lista

```python
data = [[1,'Fabian', 39], [2, 'Ana', 24], [3, 'LP', 42]]
data_from_list = pd.DataFrame(data, columns=["ID", 'Name', 'Age'])
data_from_list.head()
```

## Creación de DataFrame desde un diccionario

```python
data = [{'ID': 1, 'Name': 'José', 'Age': 38}, {'ID': 2, 'Name': 'María', 'Age': 8}]
dt_from_dict_list = pd.DataFrame(data)
print(dt_from_dict_list)
dt_from_dict_list.head()
```

```python
data = {'Id':[1,2,3,4], 'Name':['Ana','Migue','Fabian','Guillermo'], 'Age':[32,23,43,21]}
dt_from_dict = pd.DataFrame(data)
dt_from_dict.head()
```

Series

```python
data = {'ID': pd.Series([1,2,3,4]),
        'Name': pd.Series(['Ana','Migue','Fabian','Guillermo']),
        'Age': pd.Series([32,23,43,21])}
df_from_series_dict = pd.DataFrame(data)
df_from_series_dict.head()
```
