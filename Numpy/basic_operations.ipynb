# Operaciones principales en Algebra lineal
## Suma

```python
import numpy as np

A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
print(A, B)

Suma = A + B
print(Suma)
```

La multiplicación de matrices combina filas de una matriz con columnas de otra. Por ejemplo, si tenemos las mismas matrices A y B seria la fila por la columna
La multiplicación se puede realizar usando la función dot o @
A @ B
np.dot(A, B)

```python
producto = A @ B
producto_doc = np.dot(A, B)
print(producto)
print(producto_doc)
```

## Transpuesta
La transpuesta de una matríz intercambia sus filas y columnas puedo realizar la transpuesta directamente con python utilizando T
Con numpy utilizo np.transpose(matriz)

```python
A_tranpouse = A.T
print(A_tranpouse)
print("---"*10)
E = [[1,2,3],
     [4,5,6],
     [7,8,9]
    ]
#e_trans = E.T
numpy_trans = np.transpose(E)
print(numpy_trans)
```

# Determinante
El determinante es un valor único --  se calcula utilizando np.linalg.det(A)

```python
determinante = np.linalg.det(A)
print(determinante)
```

#Inversa de una matriz

```python
inversa = np.linalg.inv(A)
print(inversa)
```
