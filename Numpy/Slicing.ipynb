# Indexación y Slicing

```python
import numpy as np

arreglo = np.array([i for i in range(10, 110, 10)])
print(arreglo, f"Tamaño arreglo: {len(arreglo)}" , arreglo[-2], sep=" -- ")
print(" ---  " * 20)
print("Última posición: {}, Penúltima:{}, Del primero al tercero {}".format(arreglo[-1], arreglo[-2], arreglo[0:3]))
```

```python
# Indexación Booleana
arreglo2 = np.array([x for x in range(0, 22, 2)])
print(arreglo2)
bool_index = arreglo2 > 10
print(bool_index,end="\n ------------------------------ \n")

index_del_arreglo = [2,4,10,1]
print(arreglo2[index_del_arreglo])
```

```python
array3 = np.random.randint(1,20, size=(3,3))
print(array3)
print(array3[0,2], end="\n ============= \n")
print(array3[:2,:2]) # Array con las dos primeras filas y 2 columnas
print(array3[:3,:2]) # Array con las dos primeras filas y 2 columnas
```

```python
""" Ejercicio 2: Slicing de arreglos:
    Dado un arreglo de números del 1 al 20, extrae los elementos entre las posiciones 3 y 7, y luego los elementos 
    de las últimas 5 posiciones.
"""
array_num = [x for x in range(1,21)]
print(array_num)
print(array_num[3:8])
print(array_num[-1:-4:-1])
```
