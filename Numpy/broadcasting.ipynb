# Broadcasting


```python
import numpy as np
```

```python
prices = np.array([100,200,300,400,500])
discount = np.array([0.9])
prices_with_discount = prices * discount
print(prices_with_discount)
```

```python
prices2 = np.random.randint(100,500, size=(3,3))
plus = np.array([10,20,30])
prices_with_plus = prices2 + plus
print(prices2)
print(prices_with_plus)
```

```python
Matrix = np.arange(2, 6)
# Buscar si todos son mayores un número - Devuelve un boleano
print(np.all(Matrix > 1))
# Si existe alguno any
print(np.any(Matrix == 15))
```

```python
Matrix = np.arange(1, 10).reshape(3, 3)
array_a = Matrix[0:1,:].flatten()
array_b = Matrix[1:2,:].flatten()
array_c = np.array([7,8,9])
print(array_c)
concatenated_a = np.concatenate((array_a, array_b))
concatenated_b = np.concatenate((array_a, array_a))
print(concatenated_a, concatenated_b, sep="\n=================\n")
```
