## Vistas o Copias

```python
# Si se hace de estar forma se pasa por referencia y el cambio de una altera la otra
array_x = np.arange(10)
view_y = array_x[1:3]
print(array_x, view_y)
array_x[1:3] = [10, 11]
print(array_x, view_y)
view_y[1] = 15
print(array_x, view_y)

# Para evitar el paso por referencia utilizo COPY
copy_x = array_x[[1,2]]
print(array_x, view_y, copy_x)
array_x[1:3] = [12,13]
print(array_x, view_y, copy_x)

```
